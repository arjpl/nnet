```python
from collections import Counter

import numpy as np
```

## Data

```python
np.random.seed(42)
x = np.random.uniform(low=5, high=10, size = 10000)
y = [1 if i>=7.5 else 0 for i in x]
```

```python
Counter(y)
```

```python
x_train = x[:6000]
y_train = np.array(y[:6000])

x_test = x[6000:]
y_test = np.array(y[6000:])
```

## Functions

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def forward_pass(input, weight, bias):
    return sigmoid(input * weight + bias)
```

```python
def train(train_x, train_y, epochs = 5000, eta = 0.05):
    
    n = train_x.shape[0]
    np.random.seed(42)
    weight = np.random.uniform(-1, 1)
    bias = np.random.uniform(-1, 1)
    
    for epoch in range(epochs):
        loss = 0
        dC_dz = 0
        dC_dw = 0
        dC_db = 0
        for x_i, y_i in zip(train_x, train_y):
            a = forward_pass(x_i, weight, bias)
            loss_i = (a - y_i)**2
            loss+=loss_i
            
            dC_dz_i = 2*(a - y_i)*(a*(1-a))
            dC_dw_i = dC_dz_i * x_i
            dC_db_i = dC_dz_i
            
            dC_dz+=dC_dz_i
            dC_dw+=dC_dw_i
            dC_db+=dC_db_i

        dC_dz=dC_dz/n
        dC_dw=dC_dw/n
        dC_db=dC_db/n
        loss = loss/n
        
        weight = weight - eta * dC_dw
        bias = bias - eta * dC_db
        
        if (epoch+1) % 100 == 0:
            print(f"===============Epoch {epoch+1}/{epochs}===============")
            print(f"Loss: {loss}\n")
    
    return weight, bias
    
```

## Training the neuron

```python
trained_weight, trained_bias = train(x_train, y_train)
```

## Testing the trained neuron

```python
y_pred = np.zeros(y_test.shape[0])
test_size = y_pred.shape[0]
```

```python
for i in range(test_size):
    pred = forward_pass(x_test[i], trained_weight, trained_bias)
    y_pred[i] = 1 if pred>=0.5 else 0
```

## Accuracy

```python
np.sum(y_pred == y_test) / test_size
```
